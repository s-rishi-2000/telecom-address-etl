```python
def log_silver_audit(vendor_name, layer_name, rows_read, rows_passed, rows_quarantined):
    spark.sql(f"""
        INSERT INTO inlap.control.audit_log 
        VALUES (
            '{vendor_name}', 
            '{layer_name}', 
            current_timestamp(), 
            'SUCCESS', 
            {rows_read}, 
            {rows_passed}, 
            {rows_quarantined}
        )
    """)
    print(f"Audit Logged -> Vendor: {vendor_name} | Layer: {layer_name} | Read: {rows_read} | Passed: {rows_passed} | Quarantined: {rows_quarantined}")
```
